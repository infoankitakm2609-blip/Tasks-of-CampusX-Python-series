Example 1


```python
# In this we can any text is palindrome or not
def palin(text):
  if len(text) <= 1:
    print("Palindrome")
  else:
    if text[0] == text [-1]:
      return palin(text[1:-1])
    else:
      print("Not a palindrome")



palin("madam")
palin("python")
palin("naan")
```

```python
#The Rabbit Game (Fibonacci)
import time
def fib(m):
  if m == 0 or m ==1 :
    return 1
  else:
    return fib(m-2) + fib(m-1)

start= time.time()
print(fib(12))
end = time.time()
print("Total time taken for execution", end-start)

#This above code will take a lot of time for the execution for higher number(>40) so this will not be effective
# So we have to use memoization techniques
```

```python
def memo(m,d):
  if m in d:
    return d[m]
  else:
    d[m] = memo(m-1,d) + memo(m-2,d)
    return d[m]

d={0:1,1:1}
print(memo(48,d))
```
