```python
import time

def timer(func):
  def wrapper(*args):
    start = time.time()
    func(*args)
    print("time taken by ",func.__name__,time.time()-start,'secs')
  return wrapper

@timer
def hello():
  print("hello world")
  time.sleep(2)

@timer
def display():
  print("Displaying something!")
  time.sleep(4)

@timer
def square(num):
  time.sleep(1)
  print(num**2)

@timer
def power(num,pw):
  time.sleep(2)
  print(num**pw)
hello()
display()
square(10)
power(2,3)
```
