### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

````python
# Write your code here
import math
cor_1a=int(input("enter the first coordinate:"))
cor_1b=int(input("enter the second coordinate:"))
cor_2a=int(input("enter the third coordinate:"))
cor_2b=int(input("enter the forth coordinate:"))
result=math.sqrt((cor_2a-cor_1a)**2+(cor_2b-cor_1b)**2)
print(result)

````

### Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

For example:
Input:
heads -> 4
legs -> 12
<br>
Output:
dogs -> 2
chicken -> 2





````python
# Write your code here
heads=int(input("Enter total no of heads:"))
legs=int(input("Enter total no of legs:"))
chickens=heads
chickens_leg=heads*2
remaining=legs-chickens_leg
dogs=remaining//2
chickens=heads-dogs
print("dogs:",dogs)
print("chickens:",chickens)
````

### Q7:- Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.

````python
# Write your code here
n=int(input("Enter the nth natural number:"))
if n>0:
  sum_of_sq=n*(n+1)*(2*n+1)/6
  print("Sum of Squares:",sum_of_sq)
else:
  print("Invalid Input")
````

### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python
# Write your code here
a=int(input("enter first number:"))
b=int(input("enter second number:"))
d=b-a
n=int(input("enter the nth term:"))
An=a+(n-1)*d
print("The Nth term of the series:",An)
````

### Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.



Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm

````python
# Write your code here
T_height=int(input("enter the height of the tank:"))
T_width=int(input("enter the width of the tank:"))
T_breadth=int(input("enter the breadth of the tank:"))

G_height=int(input("enter the height of the glass:"))
G_radius=int(input("enter the radius of the glass:"))

T_volume=T_height*T_width*T_breadth
G_volume=3.14*(G_radius**2)*G_height

print("Total number of the glasses of the milk:",T_volume//G_volume)


````
