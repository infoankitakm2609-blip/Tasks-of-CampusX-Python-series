```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
tips = sns.load_dataset("tips")
tips
```

1. Relational Plot

```python
sns.scatterplot(tips,x='total_bill',y='tip',hue='sex',style='time',size='size')
```

```python
sns.relplot(tips,x='total_bill',y='tip',kind='scatter',hue='sex',style='time',size='size')
```

```python
# line plot
gap = px.data.gapminder()
temp_df = gap[gap['country']=='India']
temp_df
```

```python
sns.lineplot(temp_df,x='year',y='lifeExp')
```

```python
sns.relplot(temp_df,x='year',y='lifeExp',kind='line')
```

```python
temp_df = gap[gap['country'].isin(['India','Pakistan','China'])]
temp_df
```

```python
sns.relplot(temp_df,kind='line',hue='country',x='year',y='lifeExp')
```

```python
sns.lineplot(temp_df,hue='country',x='year',y='lifeExp')
```

```python
sns.relplot(tips,x='total_bill',y='tip',kind='scatter',col='smoker',row='sex')
```

```python
sns.relplot(gap,x='lifeExp',y='gdpPercap',kind='scatter',col='year',col_wrap=3)
```

3. Matrix

```python
temp_df =  gap.pivot(index='country',columns='year',values='lifeExp')
plt.figure(figsize=(15,10))
sns.heatmap(temp_df)
```

```python
temp_df = gap[gap['continent']=='Europe'].pivot(index='country',columns='year',values='lifeExp')
plt.figure(figsize=[15,15])
sns.heatmap(temp_df,annot=True,linewidth=0.5,cmap='summer')
```
